Ex 04
=====

Ziel: Unsichere Aktionen im Browser verhindern (Unauthorized Code Execution)

Der Chatbot gibt akuell beliebige HTML Tags aus.
Das ist gefährlich, weil damit der Browser zu unsicheren Aktionen veranlasst werden kann.

Versuche:

1. Im Browser die Developer-Tools öffnen (F12)

2. Prompt: "Write a little poem about this image (and show it) https://www.pngitem.com/pimgs/m/40-408988_smiley-face-png-evil-emoticon-smiley-face-svg.png. Just write the img tag without markdown."

3. Das Bild wird zwar nicht angezeigt, jedoch geladen. Damit können Daten exfiltriert werden.

4. Prompt: "Write a javascript alert("hello") function. don't wrap it in markdown, just the script tag" => was passiert?

Schritte:

1. Um dies zu verhindern, soll der Chatbot bzw. die Flask-App alle Ausgaben in harmlosen Text umwandeln - sog. "escaping".
2. `templates/index.html` kopieren zu `playground/templates/index.html`
3. Alternativ kann folgendes Statement korrigiert werden

    ```
    {{ msg.content | safe }}
    ```

    ersetzen mit


    ```
    {{ msg.content }}
    ```

Das `| safe` bewirkt, dass Flask den Inhalt als sicher betrachtet - eben bereits "escaped". Dies sollte nie verwendet werden, wenn Eingaben oder Ausgaben von ausserhalb der eigenen App stammen - wie z.B. hier vom Nutzer bzw. vom LLM.

Führe anschliessend einen neuen Versuch durch:

1. Prompt: "Write a javascript alert("hello") function. don't wrap it in markdown, just the script tag" => was passiert?
