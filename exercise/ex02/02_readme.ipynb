Ex 02
=====

Ziel: Login hinzufügen (Grundsicherheit)

Fragen:

* Ist da App so wie sie ist sicher?
* Was fehlt?

Schritte:

1. in `playground/__init__.py` folgen Zeile hinzufügen / Kommentare entfernen

   ```
   # ex02
   from ex02.protect import protect_all_routes
   protect_all_routes(app)
   ```
2. Refresh `http://127.0.0.1:5000/` - es sollte ein Login mit Userid/Passwort erscheinen
3. Login mit alice / password


NB: 
* Die App ist dadurch nur ein wenig sicherer geworden. Was fehlt? (Tipps: https, CORS, XSS, Session-Timeout, Same-origin Policy, ...)
* Ausserdem weisst die App einige weitere Schwachstellen auf. Was? (Tipp: was passiert, wenn sich mehr als 1 Nutzer einloggt? Was passiert mit Uploads?)
